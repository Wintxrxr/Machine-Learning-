***PART A***

In highly imbalanced datasets like credit card fraud detection, accuracy is a misleading factor because it gets dominated by the majority class. In this case, 99.5% of the transactions are legitimate and only 0.5 are fraudulent.
For example:
Consider that we have 1,000,000 transations and out of these 995000 transactions are legitimate meaing that the rest 5000 are fraudulent.
Now consider a model that predicts 'Legitimate' for every transaction.
Therefore:
True Negative = 995000
True Positive = 0
False Negative = 5000
False positive = 0
Thus accuracy comes out to 99.5% by the formula
Accuracy = TN+TP/Total
from this example we can infer that despite a very high accuracy the model fails to achieve its actual objective i.e detecting fraud.

Therefore we can say that the Area Under the ROC curve (AUC) evealuates how well the model ranks fraudulent transactions higheer than the legitimate ones, independent of any specific threshhold. It considers the trade of  b/w TP rate and FP rate.
Unlike accuracy, AUC measures the model's discriminative power and reflects performence across all possble thresholds.
Therefore, AUC gives a far more meaningful evaluation for fraud detection systems than accuracy.

***PART B***

The bank can manually review only 2% of all transactions, which means the model must strictly control the FPR(False positive rate)- since false positives correspond to legitimate transactions wrongly flagged as fraud.

The ROC curve plots as follows:
X-Axis = FPR
Y-Axis = TPR

Since the investigation team can at most review 2% of the total data, we must ensure FPR_<0.02 (less than or equal to).
On the ROC curve this means the we look fot he points to the left of exactly at FPR = 0.02.
Among these we choose the one with the highest TPR.
The target point on the ROC curve is the point with the max TPR subject to FPR_< 0.02.



```python
import numpy as np
from sklearn.metrics import roc_curve

num_samples = 10000
num_fraud = int(num_samples * 0.005)
y_true = np.zeros(num_samples, dtype=int)

fraud_indices = np.random.choice(num_samples, num_fraud, replace=False)
y_true[fraud_indices] = 1

y_scores = np.random.rand(num_samples) * 0.4
y_scores[fraud_indices] = np.random.rand(num_fraud) * 0.5 + 0.5
y_scores = np.clip(y_scores, 0, 1)


fpr, tpr, thresholds = roc_curve(y_true, y_scores)

max_fpr = 0.02

valid_indices = np.where(fpr <= max_fpr)[0]

if len(valid_indices) > 0:
    best_index = valid_indices[np.argmax(tpr[valid_indices])]

    optimal_threshold = thresholds[best_index]
    optimal_fpr = fpr[best_index]
    optimal_tpr = tpr[best_index]

    print(f"Optimal Threshold: {optimal_threshold:.4f}")
    print(f"False Positive Rate: {optimal_fpr:.4f}")
    print(f"True Positive Rate: {optimal_tpr:.4f}")
elif len(fpr) > 0:
    print(f"No operating point found with FPR less than or equal to {max_fpr:.2f}.")
    min_fpr_index = np.argmin(fpr)
    print(f"Minimum FPR found: {fpr[min_fpr_index]:.4f} with TPR: {tpr[min_fpr_index]:.4f} and Threshold: {thresholds[min_fpr_index]:.4f}")
else:
    print("ROC curve could not be computed, likely due to insufficient data or constant predictions.")
```
