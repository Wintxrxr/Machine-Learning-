```python
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import matplotlib.pyplot as plt

# --- Generate Dataset ---
np.random.seed(42)
n = 5000

data = pd.DataFrame({
    'speed_variance':       np.random.normal(30, 10, n),
    'braking_frequency':    np.random.normal(5, 2, n),
    'engine_temp':          np.random.normal(90, 15, n),
    'fuel_efficiency':      np.random.normal(12, 3, n),
    'gps_deviation':        np.random.normal(2, 1, n),
    'tyre_pressure':        np.random.normal(32, 4, n),
    'load_weight':          np.random.normal(500, 150, n),
    'idle_time':            np.random.normal(20, 8, n),
    'acceleration_spikes':  np.random.normal(3, 1, n),
    'sharp_turns':          np.random.normal(8, 3, n),
    'route_complexity':     np.random.normal(5, 2, n),
    'time_of_day':          np.random.randint(0, 24, n).astype(float),
    'driver_fatigue_score': np.random.normal(4, 1.5, n),
    'road_quality_index':   np.random.normal(6, 2, n),
    'weather_severity':     np.random.normal(3, 1.5, n),
})

# Generate label with some correlation to features
risk_score = (
    0.3 * data['speed_variance'] +
    0.4 * data['braking_frequency'] +
    0.2 * data['driver_fatigue_score'] +
    0.3 * data['weather_severity'] -
    0.2 * data['road_quality_index'] +
    np.random.normal(0, 5, n)
)
data['high_risk'] = (risk_score > risk_score.median()).astype(int)

X = data.drop('high_risk', axis=1)
y = data['high_risk']

print("Dataset shape:", data.shape)
print("Feature count:", X.shape[1])
print("Class distribution:\n", y.value_counts())

```
