```python
def simulate_neuron(inputs, weights, bias):


    x1, x2 = inputs


    w1, w2 = weights


    H = (x1 * w1) + (x2 * w2) + bias


    output = max(0, H)


    return output
```

```python
inputs = (10, 85)
weights = (0.5, 0.2)
bias = -15

result = simulate_neuron(inputs, weights, bias)

print("Neuron Output:", result)
```
