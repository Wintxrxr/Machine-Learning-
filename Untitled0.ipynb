```python
import math
import collections as counter

training_data = [
     (2, 3, 'red'),
    (3, 4, 'red'),
    (6, 5, 'blue'),
    (7, 7, 'blue'),
    (8, 6, 'blue')
]

test_point = (5,5)

K = 3

def euclidean_distance(p1,p2):
  return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

distances = [] # Moved this line out of the function definition
for x, y, label in training_data:
    dist = euclidean_distance((x, y), test_point) # Corrected function name
    distances.append((dist, label))
    print(f"Distance from {test_point} to ({x}, {y}) [{label}] = {dist:.2f}")

distances.sort(key=lambda x: x[0])

k_nearest = distances[:K]
labels = [label for _, label in k_nearest]
predicted_class = counter.Counter(labels).most_common(1)[0][0] # Corrected Counter usage

print("\nK nearest neighbors (K=3):")
for d, l in k_nearest:
    print(f"Distance = {d:.2f}, Label = {l}")

print("\nPredicted class for test point", test_point, "is:", predicted_class)
```
